#### Adding the "MONTH" Coulmn to the Dataframes

#### Adding the "MONTH" Coulmn to the Seattle Dataframe


```python
# Craete a new column on both dataframes called "month" that contains the month of the date column 
# eg 1 = Jan 2 = Feb .. , 11 = Nov , 12 = Dec 
# Create a New Column for the Month where 1 = Jan 2 = Feb etc all the way to 12 = Dec Given the Date month just contains intergers and not date time formart we can just assign the month column to be the same as the date month column

month_map = {1: "Jan", 2: "Feb", 3: "Mar", 4: "Apr", 5: "May", 6: "Jun",
             7: "Jul", 8: "Aug", 9: "Sep", 10: "Oct", 11: "Nov", 12: "Dec"}

seattle["MONTH"] = seattle["DATE"].astype(int).map(month_map)
seattle.head()
```

#### Adding the "MONTH" Coulmn to the Austin Dataframe


```python
month_map = {1: "Jan", 2: "Feb", 3: "Mar", 4: "Apr", 5: "May", 6: "Jun",
             7: "Jul", 8: "Aug", 9: "Sep", 10: "Oct", 11: "Nov", 12: "Dec"}
austin["MONTH"] = austin["DATE"].astype(int).map(month_map)
austin.head()
```
