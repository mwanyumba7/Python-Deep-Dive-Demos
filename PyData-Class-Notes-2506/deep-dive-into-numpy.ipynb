### Flattening an Array 

- Removing an array from whichever dimension it is into a 1D Array 

```python
sample_array = np.array([[12, 34, 50], 
                         [23, 45, 67], 
                         [34, 56, 78]])

sample_array.shape
```

```python
flattened_array = sample_array.flatten()
flattened_array.shape
```

### Reshaping an array 


- Changing the shape of an array to a desired shape

**Rules for Reshaping an array :**

- The sum of values has to be the same as the product of the shape values you want to change into 
- Rule 2

```python
counts_of_fireworks = np.array([[1,2],
                                [3,4],
                                [5,6]])
counts_of_fireworks.shape
```

```python
new_array = counts_of_fireworks.reshape(2, 3)
new_array
```

```python
new_array_2 = counts_of_fireworks.reshape(3,3)
new_array_2
```

### Indexing and Slicing Arrays

## BroadCasting

```python
building_heights = [234, 73, 90, 67, 90]
the_sum = building_heights[0] + building_heights[1] 
the_sum
```

```python
building_heights ** 2
```

- This problem is now solved by NumPy Arrays through Broad Casting

```python
building_heights_array = np.array(building_heights)
building_heights_array
```

```python
building_heights_array * 100
```

```python
building_heights_array ** 2
```
