## User Defined Functions

- We create User defined functions using the `def` keyword

Syntax for Functions: 

`def` function_name(`paramenter(s) names`): # Parameters are also called arguments

    logic

```python
def square(value):
    """Prints the square of a value"""
    squared_value = value ** 2
    print(squared_value)

```

```python
square(23)
```

```python
def modulous(value):
    '''Returns the remainder of a value when divided by 2'''
    remainder_value = value % 2
    return remainder_value

num = modulous(23)
print(num)
```

- Passing multiple parameters

```python
def power(base, exponent):
    """ Raise the base to the exponent and returns the result"""
    raised_value = base ** exponent
    return raised_value
```

```python
result = power(10, 3)
print(result)
```

### Error Handling 

```python
def make_a_list_of_ones(lenght):
    """ Returns a list of ones of a specified leght """
    if lenght <= 0:
        raise ValueError("Invalid Lenght. Length must be a positive integer.")
    return [1] * lenght
```

```python
make_a_list_of_ones(3)
```

```python
API_KEY = '4259a769f51363a70c3ccf5a9038c070'

def get_weather(city):
    url = f"http://api.openweathermap.org/data/2.5/weather?q={city}&appid={API_KEY}&units=metric"
    print(url)
    response = requests.get(url)
    if response.status_code == 200:
        data = response.json()
        main = data['main']
        temperature = main['temp']
        weather_description = data['weather'][0]['description']
        return f"The temperature in {city} is {temperature}°C with {weather_description}."
    else:
        return "I couldn't retrieve the weather information."
```
