## Multi-Line Strings

```python
# We use multi-line strings to create a string variable that spans multiple lines. 

mokimo_recipe = """
1 . Boil the potatoes until they are soft.
2 . Mash the potatoes and add some butter, salt, and pepper to taste.
3 . Serve the mashed potatoes with your choice of protein and vegetables.
"""
```

```python
mokimo_recipe = mokimo_recipe.upper()
print(mokimo_recipe)
```

**Key Points on Multi-Line Strings**

1. Enhance Readability
2. Used in documentation
3. Used for longer text such as instructions or error messages.

#### Replacing parts of a String

`.replace(text_to_be_replaced , text_to_chage_it_to)`

```python
colour_statement[6]
```

```python
colour_statement[0:5]
```

```python
right_colour_statement = colour_statement.replace("blue", "brown", -2)
print(right_colour_statement)
```

#### Changing Cases

```python
current_top_f1_player = "Max Verstappen"

# Convert this to lowercase

current_top_f1_player = current_top_f1_player.lower()
print(current_top_f1_player)
```

```python
# Change it to Uppercase
current_top_f1_player = current_top_f1_player.upper()
print(current_top_f1_player)
```

# Data Structures In Python

## Lists in Python

- Lists store multiple values in a single variable 
    - Lists can contain any combination of data type eg `boolean` , `string`, `int`

- Lists are ordered 
    - One can use subsetting or indexing 
    - Python starts counting from **zero** for the first element

```python
fuel_prices = [90, 150, 100, 194, 60, 76]
print(fuel_prices)
```

#### Accessing Elements of a List


- We use the square brackets to subset - []
- Sub-setting syntax :
    `name_of_list[index]`

```python
fuel_prices
```

```python
fuel_prices[2]
```

```python
fuel_prices[-4]
```

#### Slicing in Lists / Accessing Multiple Elements in a List

```python
fuel_prices
```

```python
fuel_prices[2:5]
```

```python
# Get the first three fuel prices
# These are values from index zero to index two (not inclusive of index three)
fuel_prices[:3]
```

```python
# Access all the elements from the fourth index onwards
fuel_prices[3:]
```

##### Alternating Access 

```python
fuel_prices
```

```python
# Accessing Every other element in the list
fuel_prices[::2]
```

```python
# Access every 3rd element starting from the second element (index 1)
fuel_prices[1::3]
```

## Dictionaries in Python

```python
product_prices = {"Q1-001" : 28000,
                  "T1-002" : 30900}
```

```python
product_prices.keys()
```

```python
product_prices.values()
```

```python
product_prices.items()
```

```python
# Updating a Value in a Dictionary
product_prices["T1-002"] = 30000
print(product_prices)
```

```python
# Adding a new key-value pair to the dictionary
product_prices["RS-123"] = 90000
print(product_prices)
```

```python
product_prices.update({"TG-890" : 780000})
print(product_prices)
```

## Sets in Python

- Contain Unique Data 
- They are Unchangeable
    - You can add or remove but you can't change what's already there

`set = {}`

```python
mokimo_ingredients = {"Potatoes", "Pumpkin Leaves", "Magarine", "Salt", 
                      "Beans", "Peas", "Onions", "Carrots", "Pepper"}

print(mokimo_ingredients)   
```

```python
mokimo_ingredients_list = ["Potatoes", "Pumpkin Leaves", "Magarine", "Salt", 
                           "Beans", "Peas", "Onions", "Carrots", "Pepper"]
```

```python
# Changing a list to be a set
mokimo_ingredients_set_from_list = set(mokimo_ingredients_list)
print(mokimo_ingredients_set_from_list)
```

```python
print(type(mokimo_ingredients_from_list))
```

```python
print(mokimo_ingredients_from_list)
```

```python
# Sorting my set
sorted_mokimo_ingredients = sorted(mokimo_ingredients_from_list)
print(sorted_mokimo_ingredients)
```

```python
print(type(sorted_mokimo_ingredients))
```
